```python
import torch
```

```python
x = torch.arange(4.0)

x.requires_grad_(True)

x.grad

x
```

```python
y = 2 * torch.dot(x, x)

y
```

```python
y.backward()

x.grad
```

```python
x.grad.zero_()
y = x.sum()
y.backward()
x.grad
```

```python
x.grad.zero_()
y = x * x
y.sum().backward()
x.grad
```
